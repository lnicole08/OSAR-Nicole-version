```python
#single or list of files
startdir = ["MB093C x Chrimson2", "SS77383 x Chrimson2"]

baseDirectory=[]
basegenotype=[]
for n,sd in enumerate(startdir):
    dir1 = os.path.join(root, sd)
    baseDirectory.append(dir1)
    print(dir1)
    basegenotype.append(re.search(r'^.*?(?= x )', (os.path.split(dir1)[1])).group(0))
    #basegenotype.append(re.search(r'^.*?(?=-Gal4)', (os.path.split(dir1)[1])).group(0))
    
```

```python
# all files at once (ONLY IF U PLAN TO RUN ALL AT ONCE)
startdir = os.listdir(root)

baseDirectory=[]
basegenotype=[]
for n,sd in enumerate(startdir):
    dir1 = os.path.join(root, os.listdir(root)[n])
    baseDirectory.append(dir1)
    print(dir1)
    basegenotype.append(re.search(r'^.*?(?= x )', (os.path.split(dir1)[1])).group(0))
    #basegenotype.append(re.search(r'^.*?(?=-Gal4)', (os.path.split(dir1)[1])).group(0))
    
```

```python
for loc,name in zip(baseDirectory,basegenotype):
    df = osar.osar(loc,driver = name).results
    transgenicname = loc.split("\\")[-1]
    filepath = specifiedpath + "ACC Lab Dropbox\\ACC Lab\\Nicole Lee\\Data Compilation\\osar_compiled\\2025collection\\" + transgenicname + ".csv"
    df.to_csv(filepath)
    print(name)
```
